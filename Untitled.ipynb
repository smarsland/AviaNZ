```python
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.signal import convolve2d
from scipy.ndimage import gaussian_filter1d
from scipy.ndimage import gaussian_filter
from skimage.util import view_as_blocks
from scipy.stats import rankdata
```

```python
folder_path = "test/doc_split/test/data"
files = os.listdir(folder_path)
file = files[np.random.randint(0,len(files))] # 75
img = np.load(os.path.join(folder_path, file))

def apply_log(img):
    LOG_OFFSET = 1e-7
    img = img + LOG_OFFSET
    minsg = np.min(img)
    img = 10*(np.log10(img)-np.log10(minsg))
    img = np.abs(img)
    return img

plt.figure(figsize=(12,10))
plt.imshow(apply_log(img), origin="lower", interpolation="none")
plt.show()
```

```python
flat_order_row = np.argsort(img, axis=1)
ranks_row = np.empty_like(flat_order_row)
for i in range(img.shape[0]):
    ranks_row[i, flat_order_row[i]] = np.arange(img.shape[1])

ranks_row = ranks_row / (img.shape[1] - 1)

plt.figure(figsize=(12,10))
plt.imshow(ranks_row, origin="lower", interpolation="none")
plt.show()

flat_order_col = np.argsort(ranks_row, axis=0)
ranks_col = np.empty_like(flat_order_col)
for i in range(img.shape[1]):
    ranks_col[flat_order_col[:,i], i] = np.arange(img.shape[0])

ranks_col = ranks_col / (img.shape[0] - 1)

data = ranks_row * ranks_col

#data[data<0.75]=0.75

plt.figure(figsize=(12,10))
plt.imshow(data, origin="lower", interpolation="none")
plt.show()
```

```python
H, W = img.shape
sorted_pixels = np.sort(img, axis=1)
bg_pixels = sorted_pixels[:, :W//2]

# Calculate mean and variance of background pixels per frequency band
mu0 = np.mean(bg_pixels, axis=1, keepdims=True)
var0 = np.var(bg_pixels, axis=1, keepdims=True)
    
# Normalize: z-score normalization per frequency band
sg_normalized = (img - mu0) / (np.sqrt(var0) + 1e-10)

plt.figure(figsize=(12,10))
plt.imshow(apply_log(np.abs(sg_normalized)), origin="lower", interpolation="none")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import rankdata

patch_size = 10
h, w = img.shape

# Accumulators
accum = np.zeros_like(img, dtype=float)
count = np.zeros_like(img, dtype=float)

# Loop over all shifts
for di in range(patch_size):
    for dj in range(patch_size):
        # Start patches at (di, dj)
        for i in range(di, h, patch_size):
            for j in range(dj, w, patch_size):
                patch = img[i:i+patch_size, j:j+patch_size]
                flat_patch = patch.flatten()
                
                ranks = rankdata(flat_patch, method='average')
                normalized_ranks = (ranks - 1) / (flat_patch.size - 1)
                
                # Add to accumulators
                accum[i:i+patch.shape[0], j:j+patch.shape[1]] += normalized_ranks.reshape(patch.shape)
                count[i:i+patch.shape[0], j:j+patch.shape[1]] += 1

# Average over overlapping patches
new_img = accum / count

plt.figure(figsize=(12,10))
plt.imshow(new_img, origin="lower", interpolation="none", cmap='viridis')
plt.show()
```
