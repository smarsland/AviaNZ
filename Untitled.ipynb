```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.concatenate([np.random.normal(0,1,100),np.random.normal(4,1,100)])
distances = x.reshape(-1, 1) - x.reshape(1, -1)
bandwidth = 0.01 * np.std(x) * len(x) ** (-1/5) + 1e-6
weights = np.exp(-0.5 * (distances / bandwidth) ** 2)
density = np.sum(weights, axis=1) / (len(x) * bandwidth * np.sqrt(2*np.pi))
plt.scatter(x,density)
plt.show()
```
